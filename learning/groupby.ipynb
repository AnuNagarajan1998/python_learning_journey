```python
import pandas as pd

df = pd.read_excel("sales_data.xlsx")

print(df.info())
```

```python
region_revenue = df.groupby("Region")["Revenue"].sum()
average_quantity = df.groupby("Category")["Quantity"].mean()
total_revenue = df.groupby(["Region", "Category"])["Revenue"].sum()

print(region_revenue)
print()
print(average_quantity)
print()
print(total_revenue)
```

```python
summary = df.groupby("Region")["Revenue"].agg(["sum", "mean", "max"])
print(summary)
```

```python
print(df.groupby("Sales Rep").size())
print(df.groupby("Sales Rep")["Revenue"].count())
```

```python
region_total = df.groupby("Region")["Revenue"].sum()
region_total_index = df.groupby("Region")["Revenue"].sum().reset_index()

print(region_total)
print()
print(region_total_index)
```

```python
sort_ass = df.sort_values("Revenue")
sort_des = df.sort_values("Revenue", ascending = False)
sort_mix = df.sort_values(["Region", "Revenue"], ascending = [True, False])

print(sort_ass.head())
print()
print(sort_des.head())
print()
print(sort_mix.head())
```

```python
print(region_total)
print()
print(region_total_index)
print()
print(region_total.idxmax())
print()
print(region_total_index.idxmin())
```

```python

df["region_avg"] = df.groupby("Region")["Revenue"].transform("mean")

print(df.head())
print()
print(df.groupby("Region")["Revenue"].mean())
```

```python
grouped = df.groupby("Category")
electronics_only = grouped.get_group("Electronics")

print(electronics_only.head())
```

```python
df_sort = df.sort_values(["Sale Date", "Region"])

df_sort["Cumulative_revenue"] = df_sort.groupby("Region")["Revenue"].cumsum()

df_sort_region = df_sort.loc[df_sort["Region"] == "East"]

print(df_sort.head())
print()
print(df_sort_region.head())
```
