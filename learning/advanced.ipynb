```python
#Employee Payrool management

#class creation
class Employee:

    #Constructor
    def __init__(self,e_ID,e_name,basic_salary,allowance,deductions):
        self.e_ID = e_ID
        self.e_name = e_name
        self.basic_salary = basic_salary
        self.allowance = allowance
        self.deductions = deductions

    #GROSS SALARY 
    def gross(self):
        return self.basic_salary + self.allowance
    #Net_salary
    def net(self):
        return self.basic_salary - self.deductions

    #instance Methods
    def display_payslip(self):
        print("*" *10 + "EMPLOYEE PAY SLIP" + "*" *10)
        print(f"EMPLOYEE ID: {self.e_ID:>10}")
        print(f"EMPLOYEE NAME: {self.e_name:>10}")
        print(f"BASIC SALARY: {self.basic_salary:>10.2f}")
        print(f"ALLOWANCE: {self.allowance:>10.2f}")
        print(f"Deductions:{self.deductions:>10.2f}")
        print(f"GROSS SALARY: Rs.{self.gross():>10,.2f}")
        print(f"NET SALARY:Rs.{self.net():>10,.2f}")
        print("*" *40)

#Creating Object
emp1 = Employee(101,"Arun",45000,5000,3000)
emp2 = Employee(102,"Priya", 52000,7000,4500)

emp1.display_payslip()
emp2.display_payslip()
    
```
