```python
def factorial(b):
  if(b == 0 or b == 1):
      return 1
  else:
        return b*factorial(b-1)
a = factorial(6)
print(a)
```

```python
def fibonaaci(number):
    result =[]
    a=c=0
    b=1
    result.append(0)
    for i in range(0,number):
        c=a
        a=a+b
        b=c
        result.append(a)
    return result

output = fibonacci(15)
print(output)
```

```python
def generate_prime(a):
    for i in range(1,a):
        k = 1
        count = 0
        while k != a:
            if a%k == 0:
                count += 1
            k += 1
        if count == 1:
            yield i


a = int(input("Enter the number: "))
for i in generate_prime(a):
    print(i)
```
