```python
a=input("Enter the string")
b="aeiouAEIOU"
for i, ch in enumerate(a):
    result = "vowel" if ch in b else "consonant"
    print(result)
```

```python
a=input("Enter the string")
b="aeiouAEIOU"
count = 0
for i in range(0,len(a)):
    for j in range(0,len(b)):
        if a[i] in b[j]:
            count += 1
    if count == 1:
        print("vowel")
    else:
        print("consonant")
    count = 0

    
    
```
