```python
#Using a while loop with break and continue, write a program that asks the user to guess a number between 1 and 10
#(assume the secret number is 7), and keeps asking until the correct guess is entered.
#Skip printing a message for guesses below 1 or above 10 using continue.
while True:
  a = int(input("Enter a number from 1 to 10"))
  if(a == 7):
    print("the number is found")
    break
```

```python
#Write a program that takes a list of integers and uses a for loop with if/elif/else to classify and print 
#each number as 'negative', 'zero', or 'positive'. Then use a break statement to stop processing as 
#soon as three positive numbers have been found.
numbers = [4, -2, 0, 7, -5, 3, 0, -8, 9, 6, -1, 2]
count = 0
for num in numbers:
  if num > 0:
      print(f"{num} is positive")
      count +=3
      if count == 3:
          print("3 positive numbers found")
          break
  elif num < 0:
      print(f"{num} is negative")
  else:
      print(f"{num} is zero")
    

    
```
